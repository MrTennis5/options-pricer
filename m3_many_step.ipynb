```python
import numpy as np
def call_payoff(S, K):
    return np.maximum(S - K, 0)

def binomial_call(S0, K, r, sigma, T, n):
    dt = T / n
    u = np.exp(sigma * np.sqrt(dt))
    d = 1 / u
    j = np.arange(n + 1)
    S_t = S0 * (u ** j) * (d ** (n - j))
    f = call_payoff(S_t, K)
    p = ((np.exp(r * dt)) - d) / (u - d)
    disc = np.exp(-r * dt)
    for _ in range(n):
        f = disc * (p * f[1:] + (1 - p) * f[:-1])
    return f[0]

binomial_call(S0=100, K=100, r=0.05, sigma=0.2, T=1, n=100000)
```

## American Put

```python
import numpy as np
def put_payoff(S, K):
    return np.maximum(K - S, 0)
def binomial_american_put(S0, K, r, sigma, T, n):
    dt = T / n  
    u = np.exp(sigma * np.sqrt(dt))
    d = 1 / u
    j = np.arange(n + 1)
    S_t = S0 * (u ** j) * (d ** (n - j))
    f = put_payoff(S_t, K)
    p = ((np.exp(r * dt)) - d) / (u - d)
    disc = np.exp(-r * dt)
    for i in range(n - 1, -1, -1):
        f = disc * (p * f[1:] + (1 - p) * f[:-1])
        j = np.arange(i + 1)
        S_i = S0 * (u ** j) * (d ** (i - j))
        f = np.maximum(f, put_payoff(S_i, K))
    return f[0]

binomial_american_put(100, 100, 0.05, 0.2, 1, 1000)
```
