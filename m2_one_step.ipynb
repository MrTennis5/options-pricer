```python
import numpy as np
```

```python
def call_payoff(S, K):
    return np.maximum(S - K, 0)
```

```python
def one_step_call(S0, K, u, d, r, T):
    S_u = S0 * u
    S_d = S0 * d
    f_u = call_payoff(S_u, K)
    f_d = call_payoff(S_d, K)
    p = ((np.exp(r * T)) - d) / (u - d)
    f = (np.exp(-r * T)) * (p * f_u + (1 - p) * f_d)
    return f 
```

```python
one_step_call(10, 10, 1.2, 0.8, 0.04, 0.25)
```

```python
one_step_call(S0=20, K=21, u=1.1, d=0.9, r=0.12, T=0.25)
```
