```python
import numpy as np
import matplotlib.pyplot as plt
def call_payoff(S, K):
    return np.maximum(S - K, 0)

def monte_carlo_simulation(S0, K, r, sigma, T, n):
    rng = np.random.default_rng(42)
    Z = rng.standard_normal(n)
    S_t = S0 * np.exp((r - ((sigma ** 2) / 2)) * T + sigma * np.sqrt(T) * Z)
    disc = np.exp(-r * T)
    f = np.mean(call_payoff(S_t, K)) * disc
    return f
print(monte_carlo_simulation(100, 100, 0.05, 0.2, 1, 1000000))
```
