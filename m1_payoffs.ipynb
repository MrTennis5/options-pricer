```python
import numpy as np
import matplotlib.pyplot as plt
# Payoff of a European call at expiry: max(S - K, 0)
def call_payoff(S, K):
    return np.maximum(S - K, 0)

# Payoff of European put at expiry: max(K - S, 0)
def put_payoff(S, K):
    return np.maximum(K - S, 0)
```

```python
print(call_payoff(90, 100))
print(call_payoff(130, 100))

print(put_payoff(90, 100))
print(put_payoff(130, 100))
```

```python
# makes an array of three possible stock prices at expiry
prices = np.array([90, 100, 130]) 
# Subtracts the Strike from each price at once, which gives S - K, [-10, 0, 30]
print(prices - 100)
# Sends prices to call_payoff, decides whether that value or zero is larger and returns the result
print(call_payoff(prices, 100))
print(put_payoff(prices, 100))
```

```python
S = np.linspace(0, 200, 201)
plt.plot(S, call_payoff(S, 100), label = "Call Payoff")
plt.plot(S, put_payoff(S, 100), label = "Put Payoff")
plt.title("")
plt.legend()
plt.ylabel("Payoff")
plt.xlabel("Stock Price at Expiry")
plt.show()
```
